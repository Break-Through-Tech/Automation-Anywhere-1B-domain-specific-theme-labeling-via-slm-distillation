```python
# ── CELL 5: Set HuggingFace cache to Google Drive ─────────────────────────────
# This means model weights download once and persist across sessions.
import os
from google.colab import userdata

DRIVE_ROOT = '/content/drive/MyDrive/slm-distillation'
os.environ['HF_HOME'] = f'{DRIVE_ROOT}/hf_cache'
os.makedirs(os.environ['HF_HOME'], exist_ok=True)
print(f"HF cache set to: {os.environ['HF_HOME']}")
```

```python
# ── CELL 6: Load API keys from Colab Secrets ──────────────────────────────────
# Keys are read from the 🔑 Secrets panel — never hardcoded here.
import os
from google.colab import userdata

os.environ['ANTHROPIC_API_KEY'] = userdata.get('ANTHROPIC_API_KEY')
os.environ['HF_TOKEN']          = userdata.get('HF_TOKEN')

try:
    os.environ['OPENAI_API_KEY'] = userdata.get('OPENAI_API_KEY')
except Exception:
    print('OPENAI_API_KEY not set — OK for Phase 1 (Claude is the teacher).')

print('API keys loaded from Secrets.')
```

```python
# ── CELL 7: Clone or pull the GitHub repo ─────────────────────────────────────
# *** CHANGE BRANCH_NAME to your pair's branch before running ***
import os
import subprocess
from google.colab import userdata

BRANCH_NAME = 'pair-X/model-name'   # <-- CHANGE THIS
REPO_URL = (
    f"https://{userdata.get('GITHUB_PAT')}@github.com/"
    "Break-Through-Tech/"
    "Automation-Anywhere-1A-domain-specific-theme-labeling-via-slm-distillation.git"
)
PROJECT_DIR = '/content/project'

if not os.path.exists(PROJECT_DIR):
    print('Cloning repository...')
    subprocess.run(['git', 'clone', REPO_URL, PROJECT_DIR], check=True)
else:
    print('Repository already cloned — pulling latest changes...')

os.chdir(PROJECT_DIR)
subprocess.run(['git', 'config', 'user.email', 'student@cornell.edu'], check=True)
subprocess.run(['git', 'config', 'user.name',  'Student Name'],        check=True)
subprocess.run(['git', 'checkout', BRANCH_NAME],                        check=True)
subprocess.run(['git', 'pull', 'origin', BRANCH_NAME],                  check=True)

print(f'\nReady on branch: {BRANCH_NAME}')
print(f'Working directory: {os.getcwd()}')
```
